#### A. WORD ROTATION PROBLEM

**Word Rotation Problem** moves the first letter of a given word to the end while keeping the rest of the letters in the same order.

```python
def rotate_word (text): # creates a function that receives a string called "text"
    return text [1:] + text [0]
```

```python
# Examples
print (rotate_word ("python"))
print (rotate_word ("logic"))
print (rotate_word ("code"))
print (rotate_word ("A"))
```

#### B. USERNAME BUILDER PROBLEM

**Username Builder Function** converts the names to lowercase, removes all spaces, and combines the first and last names with a period between them.

```python
def make_username (first_name, last_name):

   # changes all letters of the first name to lowercase and removes any spaces found in first name.
    first_name = first_name.lower()
    first_name = first_name.replace(" ","")

   # changes all letters of the last name to lowercase and removes any spaces found in last name.
    last_name = last_name.lower()
    last_name = last_name.replace(" ","")

    return first_name + "." + last_name # combines the first and last names using a period/
```

```python
# Examples
print (make_username ("Ada", "Lovelace"))
print (make_username ("Alan", "Turing"))
print (make_username ("Ana Maris", "De Leon"))
```

#### C. BOOKEND SWAP PROBLEM

**Bookend Swap Function** exchanges the first and last elements of a list while keeping all the middle elements in their original order.

```python
def swap_bookends(items): # creates the function named swap_bookends() that accepts a list called "items".
    first, *middle, last = items # separates the list into the first, middle, and last elements.

    return [last] + middle + [first] # returns a new list with the first and last elements switched while keeping the middle unchanged.
```

```python
# Examples
print (swap_bookends ([1, 2, 3, 4, 5, 6]))
print (swap_bookends (["red", "green", "blue"]))
print (swap_bookends ([8, 3]))
```
